```python
import pandas as pd
import requests as req
import numpy as ny
import matplotlib.pyplot as graph_plot
```

```python
from_date = input("enter from year and month hypen separated:")
to_date = input("enter to year and month hypen separated:")

power_installed_capacity_url = f"https://api.ember-energy.org/v1/installed-capacity/monthly?entity_code=IND&is_aggregate_entity=false&start_date={from_date}&end_date={to_date}&is_aggregate_series=false&include_all_dates_value_range=false&api_key=088df4ed-46ea-a471-791b-22fdc4eeea29"
response = req.get(power_installed_capacity_url).json()
print(response['data'])
power_capacity_installed_df = pd.DataFrame(response['data'])
print(power_capacity_installed_df[['entity','entity_code','date','series', 'capacity_gw']])
```

```python
year_set = set()
for r in power_capacity_installed_df.itertuples():
        year_set.update(pd.DatetimeIndex(power_capacity_installed_df['date']).year)
power_source = input("Enter source of power: ")

for year in year_set:
        installed_power_capacity = power_capacity_installed_df.loc[(power_capacity_installed_df['series'] == power_source) & (pd.DatetimeIndex(power_capacity_installed_df['date']).year == int(year))]
        print(installed_power_capacity[['series', 'capacity_gw']])
```

```python
power_capacity_graph = graph_plot.figure()
power_capacity_graph.set_figwidth(13)

x_axis_data_points = ny.array(installed_power_capacity['date'])
y_axis_data_points = ny.array(installed_power_capacity['capacity_gw'])

graph_plot.xlabel('time-period')
graph_plot.ylabel( power_source+' installed capacity (gigawatt)')

graph_plot.plot(x_axis_data_points, y_axis_data_points, marker = 'x')
#graph_plot.bar(x_axis_data_points, y_axis_data_points, color='green')
graph_plot.plot()
graph_plot.grid()
```
